**Typical `_quarto.yml` structure for a Quarto book with both HTML and PDF outputs + Bootstrap PDF download icon in the right-hand sidebar:**

```yaml
project:
  type: book

book:
  title: "Your Book Title"
  author: "Your Name"
  date: "today"   # or a specific date
  chapters:
    - index.qmd
    - intro.qmd
    - chapter1.qmd
    # ... your other chapters

  # This adds the download buttons (with Bootstrap icons) in the sidebar
  downloads: [pdf]          # or [pdf, epub] etc.
  # Optional: other sidebar tools
  # repo-url: https://github.com/your/repo
  # repo-actions: [edit]
  # sharing: [twitter, facebook]

format:
  html:
    theme: cosmo          # or any Bootstrap theme you like
    toc: true
    code-download: true   # optional: code download button
    # Other HTML options...

  pdf:
    documentclass: book   # or scrbook, article, etc.
    classoption: [openany]
    toc: true
    number-sections: true
    # Other PDF/LaTeX options (geometry, fontsize, etc.)
```

### How the PDF icon appears

- `downloads: [pdf]` under the `book:` key automatically generates a **PDF download button** (with the standard Bootstrap PDF icon) in the HTML book's sidebar, usually right below the title/search area.

- It links to the rendered `Your-Book-Title.pdf` that Quarto produces alongside the HTML.

### Quick tips

- Run `quarto render` (or `quarto preview`) → it builds **both** HTML and PDF versions.
- The PDF uses Pandoc + LaTeX (TinyTeX), so make sure you have it installed.
- Want EPUB too? Change to `downloads: [pdf, epub]`.
- Customize the sidebar further with `book: sidebar: ...` (tools, search, etc.).
- For more control over the PDF look, add options under `pdf:` (e.g., `include-in-header`, custom `.tex` templates, etc.).

This is the standard/official way shown in the Quarto book documentation. Let me know if you want a more customized sidebar (e.g., right-side tools) or specific PDF styling!
